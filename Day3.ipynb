```python
# Import NumPy library
import numpy as np

# Import Pandas library
import pandas as pd


# ---------------- Dictionary to DataFrame ----------------

# Dictionary where keys become column names
# and lists become column values
data = {
    'Name': ["Narinder", "Kali", "Lalita"],
    'age': [21, 55, 67],
    'class': ["7th", "9", "10"]
}

# Convert dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Display the first 5 rows of the DataFrame
# Here, only 3 rows are available, so all 3 will be displayed
df.head()


# ---------------- List of Dictionaries to DataFrame ----------------

# Each dictionary represents one row
# Keys represent column names
listval = [
    {'Name': "Ram", 'Class': 5, 'City': "Mohali"},
    {'Name': "Mohan", 'Class': 7, 'City': "Muradbad"},
    {'Name': "Abhi", 'Class': 2, 'City': "Nangal"},
    {'Name': "Rajan", 'Class': 4, 'City': "Ropar"},
    {'Name': "Sham", 'Class': 3, 'City': "Ludhiana"}
]

# Convert list of dictionaries into a DataFrame
df = pd.DataFrame(listval)

# Display the complete DataFrame
df
```

```python
# Create a NumPy 2D array
# Each inner list represents one row
a = np.array([
    ["Tom", 24, "Tri"],
    ["Mat", 23, "Tier"],
    ["Camt", 14, "Tqwr"]
])

# Convert the NumPy array into a Pandas DataFrame
# columns parameter assigns names to each column
df = pd.DataFrame(
    a,
    columns=["Name", "Age", "City"]
)

# Display the DataFrame
df
```

```python
# Create a Pandas Series for student IDs
id = pd.Series([211, 222, 233, 244])

# Create a Pandas Series for student names
name = pd.Series(["Mon", "Raj", "Kartik", "Kom"])

# Create a Pandas Series for city names
city = pd.Series(["RajNagar", "TilakNagar", "TriCity", "ViharNagar"])

# Display the Name Series
print(name)

# Combine multiple Series into a DataFrame
# Dictionary keys become column names
# Each Series becomes one column
df = pd.DataFrame({
    'Id': id,
    'Name': name,
    'City': city
})

# Display the DataFrame
df
```

```python
# Create a dictionary containing student information
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Narinder", "Rahul", "Aman", "Simran", "Karan"],
    "Age": [21, 22, 20, 21, 23],
    "Marks": [85, 90, 78, 92, 67],
    "Department": ["CSE", "IT", "CSE", "AI", "IT"]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Save the DataFrame as a CSV file
# index=False prevents Pandas from saving the DataFrame index as an extra column
df.to_csv("students.csv", index=False)
```

```python
# Import Pandas
import pandas as pd

# Create a dictionary containing student information
data = {
    "Name": ["Narinder", "Rahul", "Aman", "Simran"],
    "Age": [21, 22, 20, 21],
    "Marks": [85, 90, 78, 92],
    "City": ["Mohali", "Nangal", "Chandigarh", "Ludhiana"]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Export the DataFrame to an Excel file
# index=False prevents the DataFrame index from being saved as an extra column
df.to_excel("students.xlsx", index=False)
```
